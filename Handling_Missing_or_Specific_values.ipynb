```python
#np.isnan()  finding mising values
import numpy as np
arr=np.array([10,np.nan,20,30,40,np.nan,80])
print(np.isnan(arr))
```

```python
#np.nan_to_num(arr,nan=value)
import numpy as np
arr=np.array([10,np.nan,20,30,40,np.nan,80])
cleaned_arr=np.nan_to_num(arr,nan=130)
print(cleaned_arr)
```

```python
#np.isinf(arr)
import numpy as np
arr=np.array([10,np.inf,20,30,40,np.inf,80])
print(arr)
print(np.isinf(arr))

```

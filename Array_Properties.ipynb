```python
#To know shape of the array
import numpy as np 
array=np.array([[1,2,3],[5,4,8]])
print(array)
print(array.shape)
```

```python
#To know size of the array
import numpy as np 
array=np.array([[1,2,3],[5,4,8]])
print(array)
print(array.size)
```

```python
#To know symention of the array
import numpy as np 
array1=np.array([1,2,3])
array2=np.array([[1,2,3],[5,4,8]])
array3=np.array([[[1,2,3],[5,4,8],[7,8,9]]])

print(array1.ndim)
print(array2.ndim)
print(array3.ndim)
```

```python
#To check the datatype of the array
import numpy as np 
array=np.array([[1,2,3],[5,4,8]])
print(array)
print(array.dtype)
```

```python
#To change the datatype of the array
import numpy as np 
array=np.array([1.2,2.5,8.9])
int_arr=array.astype(int)
print(int_arr)
print(int_arr.dtype)
```

```python
#Array filled with zeros
import numpy as np 
array_zeros=np.zeros((3,4))
print(array_zeros)
```

```python
#Array filled with ones
import numpy as np 
array_ones=np.ones((4,4))
print(array_ones)
```

```python
#An array with a specific range or sequence of elements
import numpy as np 
array_range=np.arange(1,10,2)
print(array_range)
```

```python
#For creating a empty array
import numpy as np 
empty_array=np.empty(5)
print(empty_array)
```

```python
#An array for diagonal amtrix or identity matrix
import numpy as np
identity_matrix=np.eye(3,3)
print(identity_matrix)
```

```python
#Array with a specific element
import numpy as np 
specific_array=np.full((3,5),7)
print(specific_array)
```
